```python
import json
import spacy
from spacy.matcher import PhraseMatcher
```

```python
# Load spaCy blank model for tokenization alignment checking
nlp = spacy.blank("en")
```

```python
# Load skills from file
with open("../data/linkedin_skills.txt", "r", encoding="utf-8") as f:
    skills = [line.strip() for line in f if line.strip()]
```

```python
# Set up PhraseMatcher
matcher = PhraseMatcher(nlp.vocab, attr="LOWER")
patterns = [nlp.make_doc(skill) for skill in skills]
matcher.add("SKILL", patterns)
```

```python
# Load job description sentences (assume list of strings)
with open("../data/job_sentences.json", "r", encoding="utf-8") as f:
    job_sentences = json.load(f)
```

```python
# Generate training data
training_data = []
for sentence in job_sentences:
    doc = nlp(sentence)
    matches = matcher(doc)
    
    entities = []
    for match_id, start, end in matches:
        span = doc[start:end]
        entities.append((span.start_char, span.end_char, "SKILL"))

    if entities:
        training_data.append((sentence, {"entities": entities}))
```

```python
# Save formatted training data
with open("../data/temp.json", "w", encoding="utf-8") as f:
    json.dump(training_data, f, indent=4)

print("Training data created and saved to 'temp.json'")
```

```python

output_file = "../data/spacy_training_data.json"
with open(output_file, "w") as f:
    json.dump(training_data, f, indent=4)
```
